```python
#listedeki index sayısına bakmanın kolay yolu
my_list = range(10,25)
for ix in range(len(my_list)):
    print(ix)
```

```python
#listedeki index sayısına bakmanın kolay yolu ise burada
#enumerate
for (element) in enumerate(my_list):
    print(element)
    
```

```python
#Birden fazla yapıyı tek bir yapı haline getirir
food_list = ["apple","banana","melon"]
calories_list = [100,200,300]    

zipped_list = list(zip(food_list,calories_list))
zipped_list 
```

```python
new_list = []
my_string = "metallica"

for element in my_string:
    new_list.append(element)
print(new_list)    
```

```python
#ya da bunun yerine şunu da kullanabiliriz
#list comprehesion

new_list = [element for element in my_string]
new_list
```
