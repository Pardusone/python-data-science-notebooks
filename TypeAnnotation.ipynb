```python
def addnumbers(x : int, y:int) -> int : #bu da bir gösterim x int olacak y int olacak ve yine int türünde bir veri return edilecek demek
    return x + y 
```

```python
addnumbers(3,4)
```

```python
#python 3.10 ve üzerinde var
def process_value(value: int | str)-> str: #girdi int yada str olacak ama str return edilecek demek
    if isinstance(value , int): #değerin türünü kontrol ediyor
        return f"processed integer : {value}"
    else:
        return f"processed string : {value}"
        
```

```python
process_value(50)
```

```python
#python 3.9 ve üzerinde var

from typing import List

def sum_list(numbers : List[int]) -> int: #bu fonksiyon bir intger listesi alacak ve integer return edecek demek
    return sum(numbers)
    
```

```python
numbers = [1,2,3,4]
sum_list(numbers)
```
