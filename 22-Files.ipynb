```python
with open("myfile.txt") as myFile :
    myContent = myFile.read() #burada işlemler kolaylaşsın diye atama yaptık
    
```

```python
myContent
```

```python
with open("myfile.txt",mode = "w") as myNewFile:#burada modu write yaptık içerik komple değişti
    myNewFile.write("test 4") #burada da ne yazmak istiyorsak onu ekledik 
```

```python
myNewFile 
```

```python
with open("myfile.txt",mode = "r") as myFile2: #burada da okuma görevini bir değişkene atadık
    myContent = myFile2.read()
```

```python
myContent
```

W -> write , r -> read , a -> append

```python
with open("myfile.txt",mode = "a") as myNewFile3:
    myNewFile3.write("test 5")    
```

```python
with open("myfile.txt",mode = "r") as myNewFile4:
    myContent = myNewFile4.read()
```
