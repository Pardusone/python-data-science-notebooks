```python
def divideNumber(number):
    return number / 2 
divideNumber(20)
```

```python
my_list = [3,5,7,10,20,30]
```

```python
my_result_list = []

for i in my_list:
    my_result_list.append(divideNumber(i))
my_result_list
```
