```python
#args , kwargs (arguements , key word arguments)
```

```python
def args_sum(*args) :
    return sum(args)
args_sum(10,20,30,40) # istediğin kadar yazabilirsin
```

```python
def kwargs_example(**kwargs): #key word arguement yazabilirsin yani sözlük
    print(kwargs)

kwargs_example(apple = 100 , banana = 150 , melon = 200)
```
