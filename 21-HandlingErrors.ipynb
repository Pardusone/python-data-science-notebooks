```python
while True:
    try:
        myAge = int(input("Enter age:"))
        print(myAge*2)
        break
    except ValueError:
        print("Enter your age!!")
    else:
        print("Else executed")
    finally: #finally her zaman çalışır break önemsiz
        print("finally")
```
